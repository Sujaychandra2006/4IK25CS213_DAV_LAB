```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("/content/student_performance_dataset (2).csv")
print("First 5 rows of dataset:")
print(df.head)
```

```python
print("\nMissing values before handling:")
print(df.isnull().sum())
```

```python
df.loc[0:5,'Study-Hours_per_week'] = None
df.loc[10:15, 'Attendemce_Rate'] = None
df.loc[20:25, 'Past_Exam_Scores'] = None
df.loc[30:35, 'Final_Exam_Score'] = None
print("\nMissing values after inserting sample NaNs")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
df['Study_Hours_per_Week'] = df['Study_Hours_per_Week'].fillna(df['Study_Hours_per_Week'].mean())
```

```python
df['Attendance_Rate'] = df['Attendance_Rate'].fillna(df['Attendance_Rate'].median())
```

```python
df['Past_Exam_Scores'] = df['Past_Exam_Scores'].ffill()
```

```python
df['Final_Exam_Score'] = df['Final_Exam_Score'].fillna(df['Final_Exam_Score'].mean())
print("\nMissing values after imputation:")
print(df.isnull().sum())
```

```python
df = pd.get_dummies(df, columns=['Parental_Education_Level'])
print("\nOne-Hot Encoded columns:")
print(df.columns.tolist())
```

```python
ls = LabelEncoder()
df['Pass_Fail_Encoded'] = ls.fit_transform(df['Pass_Fail'])
print("\nLabel Encoded Result:")
print(df[['Pass_Fail','Pass_Fail_Encoded']].head())
```

```python
df['Gender_Encoded'] = ls.fit_transform(df['Gender'])
df['Internet_Access_Encoded'] = ls.fit_transform(df['Internet_Access_at_Home'])
df['Extracurricular_Activities_Encoded'] = ls.fit_transform(df['Extracurricular_Activities'])
```

```python
df.to_csv("student_performance_cleaned.csv", index=False)
print("\nCleaned dataset saved as 'student_performance_cleaned.csv'")
```
